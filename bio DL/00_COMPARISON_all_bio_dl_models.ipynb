# Bio-Inspired Deep Learning Models - Comprehensive Comparison

This notebook compares all **12 bio-inspired deep learning models** for edge detection from Table 2.

**Models evaluated:**
1. XYW-Net - Parallel X/Y/W cell pathways
2. LVP-Net - Lateral Visual Pathway with feedback
3. BFCN - Bio-inspired Feature Contextual Network
4. MI-Net - Multi-scale Integration Network
5. BLCDNet - Biomimetic Lightweight Contour Detection
6. DPED - Deep Pathway with attention mechanisms
7. BESD - Bio-inspired End-Stopped edge Detection
8. LEDNet - Lightweight with magno/parvo pathways
9. Bio-IFN - Interactive Feedback Network
10. BFE-Net - Bio-inspired Feature Enhancement
11. SNN-EdgeNet - Spiking Neural Network
12. Tang et al. - nCRF for Contour Detection

**Evaluation Metrics:** ODS, OIS, AP on HED_Small test set

```python
from pathlib import Path
import subprocess, sys
subprocess.run([sys.executable, '-m', 'pip', 'install', '-q', 'matplotlib', 'seaborn', 'pandas'], check=False)
import json, pandas as pd, matplotlib.pyplot as plt, seaborn as sns

OUTPUT_DIR = Path('..') / 'bio DL' / 'outputs'
```

## Load All Results

```python
models_info = [
    ('XYW-Net', 'outputs/XYW-Net/xywnet_metrics.json'),
    ('LVP-Net', 'outputs/LVP-Net/lvpnet_metrics.json'),
    ('BFCN', 'outputs/BFCN/bfcn_metrics.json'),
    ('MI-Net', 'outputs/MI-Net/minet_metrics.json'),
    ('BLCDNet', 'outputs/BLCDNet/blcdnet_metrics.json'),
    ('DPED', 'outputs/DPED/dped_metrics.json'),
    ('BESD', 'outputs/BESD/besd_metrics.json'),
    ('LEDNet', 'outputs/LEDNet/lednet_metrics.json'),
    ('Bio-IFN', 'outputs/Bio-IFN/bioifn_metrics.json'),
    ('BFE-Net', 'outputs/BFE-Net/bfenet_metrics.json'),
    ('SNN-EdgeNet', 'outputs/SNN-EdgeNet/snnedgenet_metrics.json'),
    ('Tang nCRF', 'outputs/Tang-nCRF/tang_ncrf_metrics.json'),
]

# Create output folders if they don't exist
for name, path in models_info:
    json_path = OUTPUT_DIR.parent / path
    json_path.parent.mkdir(parents=True, exist_ok=True)

print("✓ Output folders verified/created\n")

results = []
for name, path in models_info:
    json_path = OUTPUT_DIR.parent / path
    if json_path.exists():
        with open(json_path) as f:
            data = json.load(f)
            results.append({
                'Model': data.get('model', name),
                'Bio-Inspiration': data.get('bio', 'N/A'),
                'Improvement': data.get('improvement', 'N/A'),
                'ODS': data['metrics']['ODS'],
                'OIS': data['metrics']['OIS'],
                'AP': data['metrics']['AP']
            })
    else:
        print(f"⚠️ Missing: {name} ({json_path})")

df = pd.DataFrame(results)
print(f"\n✓ Loaded {len(df)} model results")
if len(df) > 0:
    print(df.to_string(index=False))
else:
    print("\n⚠️ No metrics files found. Run individual model notebooks to generate results.")
```

## Performance Ranking

```python
# Rank by ODS
df_ranked = df.sort_values('ODS', ascending=False).reset_index(drop=True)
df_ranked['Rank'] = df_ranked.index + 1

print("\n🏆 Model Ranking by ODS:")
print(df_ranked[['Rank', 'Model', 'ODS', 'OIS', 'AP', 'Bio-Inspiration']].to_string(index=False))
```

## Metrics Comparison

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

metrics = ['ODS', 'OIS', 'AP']
colors = ['#3498db', '#e74c3c', '#2ecc71']

for ax, metric, color in zip(axes, metrics, colors):
    sorted_df = df.sort_values(metric, ascending=True)
    ax.barh(sorted_df['Model'], sorted_df[metric], color=color, alpha=0.7)
    ax.set_xlabel(metric, fontsize=12, fontweight='bold')
    ax.set_title(f'{metric} Comparison', fontsize=14, fontweight='bold')
    ax.grid(axis='x', alpha=0.3)
    for i, v in enumerate(sorted_df[metric]):
        ax.text(v + 0.005, i, f'{v:.4f}', va='center', fontsize=9)

plt.tight_layout()
plt.savefig(OUTPUT_DIR / 'comparison_metrics.png', dpi=300, bbox_inches='tight')
print("\n✓ Saved: comparison_metrics.png")
plt.show()
```

## Statistical Summary

```python
stats = df[['ODS', 'OIS', 'AP']].describe()
print("\n📊 Statistical Summary:")
print(stats)

print(f"\n🥇 Best ODS: {df.loc[df['ODS'].idxmax(), 'Model']} ({df['ODS'].max():.4f})")
print(f"🥇 Best OIS: {df.loc[df['OIS'].idxmax(), 'Model']} ({df['OIS'].max():.4f})")
print(f"🥇 Best AP: {df.loc[df['AP'].idxmax(), 'Model']} ({df['AP'].max():.4f})")
```

## Bio-Inspiration Categories

```python
print("\n🧬 Bio-Inspiration Summary:")
for _, row in df.iterrows():
    print(f"• {row['Model']:15s} → {row['Bio-Inspiration']:40s} | {row['Improvement']}")
```

## Correlation Analysis

```python
fig, ax = plt.subplots(figsize=(8, 6))
corr = df[['ODS', 'OIS', 'AP']].corr()
sns.heatmap(corr, annot=True, fmt='.3f', cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8}, ax=ax)
ax.set_title('Metric Correlation Matrix', fontsize=14, fontweight='bold')
plt.savefig(OUTPUT_DIR / 'correlation_matrix.png', dpi=300, bbox_inches='tight')
print("\n✓ Saved: correlation_matrix.png")
plt.show()
```

## Export Comparison Results

```python
# Save to CSV
df_ranked.to_csv(OUTPUT_DIR / 'bio_dl_comparison_results.csv', index=False)
print("\n✅ Saved: bio_dl_comparison_results.csv")

# Save summary JSON
summary = {
    'total_models': len(df),
    'best_ods': {'model': df.loc[df['ODS'].idxmax(), 'Model'], 'value': float(df['ODS'].max())},
    'best_ois': {'model': df.loc[df['OIS'].idxmax(), 'Model'], 'value': float(df['OIS'].max())},
    'best_ap': {'model': df.loc[df['AP'].idxmax(), 'Model'], 'value': float(df['AP'].max())},
    'statistics': stats.to_dict(),
    'rankings': df_ranked.to_dict(orient='records')
}

with open(OUTPUT_DIR / 'bio_dl_comparison_summary.json', 'w') as f:
    json.dump(summary, f, indent=2)
print("✅ Saved: bio_dl_comparison_summary.json")
print("\n✅ Comparison complete!")
```
